```python
import sqlite3

conexao = sqlite3.connect('/content/vagas_tech.db')
cursor = conexao.cursor()

for tabela in ['VAGAS', 'CANDIDATAS', 'CANDIDATURAS']:
    cursor.execute(f"SELECT COUNT(*) FROM {tabela}")
    quantidade = cursor.fetchone()[0]
    print(f"{tabela}: {quantidade} registro(s)")

conexao.close()
```

```python
import sqlite3

conexao = sqlite3.connect('/content/vagas_tech.db')
cursor = conexao.cursor()

print("=== VALIDAÇÃO FINAL ===")

cursor.execute("""
SELECT ID_VAGA, TITULO, EMPRESA, SALARIO
FROM VAGAS;
""")

print("\nVAGAS:")
for registro in cursor.fetchall():
    print(registro)

cursor.execute("""
SELECT ID_CANDIDATA, NOME, EMAIL
FROM CANDIDATAS;
""")

print("\nCANDIDATAS:")
for registro in cursor.fetchall():
    print(registro)

cursor.execute("""
SELECT ID_CANDIDATURA, DATA_ENVIO, ID_VAGA, ID_CANDIDATA
FROM CANDIDATURAS;
""")

print("\nCANDIDATURAS:")
for registro in cursor.fetchall():
    print(registro)

conexao.close()
```
